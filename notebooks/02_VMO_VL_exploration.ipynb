# 02 : VMO–VL Exploration

## Objective
This notebook explores the relationship between the two EMG channels associated with the vastus medialis obliquus (VMO) and vastus lateralis (VL) during squatting.

## Workflow
1. Load the EMG dataset and inspect its dimensions and initial rows.
2. Rectify the channel signals using absolute values.
3. Calculate rolling root mean square (RMS) values using a 100-sample window.
4. Calculate and plot the ratio between the two RMS signals.
5. Assign preliminary activity categories using predefined ratio thresholds.
6. Summarise the resulting category counts.

## Interpretation note
The ratio and categories are exploratory measures. The threshold-based labels are not clinically validated diagnoses.

## Data requirement
The original EMG CSV must be available at the file path specified in the code before the notebook can run successfully.

```python
import pandas as pd

df = pd.read_csv('/content/emg_final.csv')
print(df.head())
print(df.shape)
import numpy as np

# Rectify (absolute value)
rectified = np.abs(df)

# RMS function
def rms(signal, window=100):
    return np.sqrt(np.convolve(signal**2, np.ones(window)/window, mode='same'))
rectified = np.abs(df)
df['RMS_1'] = rms(rectified['Channel_1'])
df['RMS_2'] = rms(rectified['Channel_2'])
df['ratio'] = df['RMS_2'] / (df['RMS_1'] + 1e-6)
import matplotlib.pyplot as plt

plt.figure(figsize=(10,4))
plt.plot(df['ratio'])
plt.title("Muscle Activation Ratio (VMO/VL)")
plt.xlabel("Time")
plt.ylabel("Ratio")
plt.show()

def classify(r):
    if r < 0.8:
        return "VL dominant (imbalance)"
    elif r > 1.2:
        return "VMO dominant"
    else:
        return "Balanced"

df['status'] = df['ratio'].apply(classify)

print(df['status'].value_counts())
```
